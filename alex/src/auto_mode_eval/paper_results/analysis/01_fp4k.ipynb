```python
import pandas as pd

from auto_mode_eval.paper_results.analysis.results import grid, samples

pd.set_option("display.width", 220)

cells = grid("fp4k")
fp = samples("fp4k")
print(fp.shape)
cells[["arm", "samples", "epochs", "status", "revision", "exact_mean"]]
```

```python
union = set(fp.id.unique())
check = fp.groupby("arm").agg(
    rows=("id", "size"),
    actions=("id", "nunique"),
    epochs=("epoch", "nunique"),
    # `error` is an empty string, not NA, when the sample was fine
    errors=("error", lambda s: (s.fillna("") != "").sum()),
    unscored=("score_exact", lambda s: s.isna().sum()),
)
check["missing"] = fp.groupby("arm").id.apply(lambda s: len(union - set(s)))
check["complete"] = (
    (check.rows == 12000)
    & (check.actions == 4000)
    & (check.epochs == 3)
    & (check.missing == 0)
    & (check.unscored == 0)
    & (check.errors == 0)
)

print(f"{len(union)} distinct actions; every arm covers all of them: {bool(check.complete.all())}")
check
```
